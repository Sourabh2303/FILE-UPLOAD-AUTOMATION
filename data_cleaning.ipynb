```python
def normalize_verfied(value):
    if pd.isna(value):
        return value
    value=str(value).strip().lower()
    if value in ["yes","verified","y"]:
        return "Y"
    elif value in ["no","n"]:
        return "N"
    else:
        return value

cbnexus_contacts_df["Verified"]=cbnexus_contacts_df["Verified"].apply(normalize_verfied)
cbnexus_contacts_df
```

```python
import re
def normalize_phone(value):
    if pd.isna(value):
        return value
    phone=str(value).strip()
    phone=re.sub(r"\D","",phone)

    if phone.startswith("91")and len(phone)==12:
        phone=phone[2:]
    return phone
cbnexus_contacts_df["Phone Number"]=cbnexus_contacts_df["Phone Number"].apply(normalize_phone)
cbnexus_contacts_df
```

```python
def normalize_date(value):
    if pd.isna(value):
        return pd.NaT

    value = str(value).strip()

    formats = [
        "%d-%m-%Y",   # 24-07-2026
        "%d/%m/%Y",   # 24/07/2026
        "%m/%d/%Y",   # 07/13/2026
        "%d %b %Y",   # 7 Jul 2026
        "%d %B %Y"    # 7 July 2026
    ]

    for fmt in formats:
        try:
            return pd.to_datetime(value, format=fmt)
        except (ValueError, TypeError):
            pass

    # If nothing matched, keep the original value
    return value
naukri_df["Applied Date"]=naukri_df["Applied Date"].apply(normalize_date)
naukri_df["Applied Date"] = pd.to_datetime(
    naukri_df["Applied Date"],
    format="mixed"
).dt.date
naukri_df
```

```python
combined_df["name_key"] = (
    combined_df["name"]
    .astype("string")
    .str.strip()
    .str.lower()
)
combined_df["email_key"] = (
    combined_df["email"]
    .astype("string")
    .str.strip()
    .str.lower()
)


combined_df["phone_key"] = (
    combined_df["phone"]
    .astype("string")
    .str.strip()
)
combined_df[
    ["name", "email", "phone", "source",
     "name_key", "email_key", "phone_key"]
].head(10)
```

```python
matching_df = combined_df.copy()
email_to_person = {}
phone_to_person = {}

person_counter = 1

person_ids = []

for _, row in matching_df.iterrows():

    email = row["email_key"]
    phone = row["phone_key"]
    name = row["name_key"]

    person_id = None

    # 1. Try email match
    if pd.notna(email) and email in email_to_person:
        person_id = email_to_person[email]

    # 2. If no email match, try phone
    elif pd.notna(phone) and phone in phone_to_person:
        person_id = phone_to_person[phone]

    # 3. No existing match → create new person
    else:
        person_id = f"P{person_counter:04d}"
        person_counter += 1

    # Store the assigned person ID
    person_ids.append(person_id)

    # Update mappings
    if pd.notna(email):
        email_to_person[email] = person_id

    if pd.notna(phone):
        phone_to_person[phone] = person_id


matching_df["person_id"] = person_ids
matching_df
```

```python
def first_valid(series):
    values = series.dropna()
    
    if len(values) == 0:
        return pd.NA
    
    return values.iloc[0]

def combine_skills(series):
    skills = set()

    for value in series.dropna():
        for skill in str(value).split(","):
            skill = skill.strip()

            if skill:
                skills.add(skill)

    return ", ".join(sorted(skills))


def combine_emails(series):
    emails = set()

    for value in series.dropna():
        value = str(value).strip().lower()

        if value:
            emails.add(value)

    return ", ".join(sorted(emails))
```

```python
master_people = (
    matching_df
    .groupby("person_id")
    .agg(
        name=("name", first_valid),
        email=("email", combine_emails),
        phone=("phone", first_valid),
        city=("city", first_valid),
        skills=("skills", combine_skills),
        sources=("source", lambda x: ", ".join(sorted(set(x))))
    )
    .reset_index()
)
```

```python
import sqlite3

conn = sqlite3.connect("assignment.db")
cursor = conn.cursor()

print("Database connected successfully!")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS persons (
    person_id TEXT PRIMARY KEY,
    name TEXT NOT NULL,
    email TEXT,
    phone TEXT,
    city TEXT,
    skills TEXT,
    sources TEXT
)
""")
cursor.execute("""
CREATE TABLE IF NOT EXISTS naukri_records (
    source_record_id TEXT PRIMARY KEY,
    person_id TEXT NOT NULL,
    name TEXT,
    email TEXT,
    phone TEXT,
    city TEXT,
    experience_years REAL,
    current_ctc REAL,
    applied_date TEXT,
    skills TEXT,
    FOREIGN KEY (person_id) REFERENCES persons(person_id)
)
""")
cursor.execute("""
CREATE TABLE IF NOT EXISTS gig_worker_records (
    source_record_id TEXT PRIMARY KEY,
    person_id TEXT NOT NULL,
    name TEXT,
    email TEXT,
    phone TEXT,
    city TEXT,
    rate TEXT,
    status TEXT,
    skills TEXT,
    FOREIGN KEY (person_id) REFERENCES persons(person_id)
)
""")
cursor.execute("""
CREATE TABLE IF NOT EXISTS cbnexus_records (
    source_record_id TEXT PRIMARY KEY,
    person_id TEXT NOT NULL,
    name TEXT,
    phone TEXT,
    city TEXT,
    verified TEXT,
    projects_completed INTEGER,
    FOREIGN KEY (person_id) REFERENCES persons(person_id)
)
""")
cursor.execute("""
CREATE TABLE IF NOT EXISTS person_sources (
    person_id TEXT NOT NULL,
    source TEXT NOT NULL,
    PRIMARY KEY (person_id, source),
    FOREIGN KEY (person_id) REFERENCES persons(person_id)
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS audio_submissions (
    submission_id INTEGER PRIMARY KEY AUTOINCREMENT,
    person_id TEXT,
    file_path TEXT NOT NULL,
    transcript TEXT,
    extracted_fields TEXT,
    created_at TEXT DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (person_id) REFERENCES persons(person_id)
)
""")
```
